```python
!pip install git+https://github.com/google/myanmar-tools.git#subdirectory=clients/python openpyxl pandas -q

import pandas as pd
from google.colab import files
from myanmartools import ZawgyiDetector

detector = ZawgyiDetector()

def detect(text):
    if not isinstance(text, str) or not text.strip():
        return ""
    score = detector.get_zawgyi_probability(text)
    if score == float("-inf"):
        return "Not Burmese"
    return "Zawgyi" if score >= 0.5 else "Unicode"

uploaded = files.upload()                  # choose your Excel file
filename = list(uploaded.keys())[0]
df = pd.read_excel(filename)

print("Columns:", list(df.columns))
col = "Unnamed: 0"                     # updated to the actual column with Burmese text

df["Font_Type"] = df[col].apply(detect)    # new column with the result

df.to_excel("result.xlsx", index=False)
files.download("result.xlsx")
df.head()

from google.colab import sheets
sheet = sheets.InteractiveSheet(df=df)
```
